```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import local_binary_pattern

# Function to display images
def show_image(title, img, cmap=None):
    plt.figure(figsize=(6, 6))
    plt.title(title)

    if cmap:
        plt.imshow(img, cmap=cmap)
    else:
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

    plt.axis('off')
    plt.show()


# 1. Load the Image
# Replace with your image filename
img = cv2.imread('/content/drive/MyDrive/COMPUTER VISION/image16.jpg')

# Check whether image is loaded successfully
if img is None:
    raise FileNotFoundError("Image not found. Check the image filename and path.")

# 2. Pre-processing
# Convert the image into grayscale
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# 3. LBP Implementation
# Define LBP parameters
radius = 1
n_points = 8 * radius

# Calculate Local Binary Pattern
lbp = local_binary_pattern(
    gray_img,
    n_points,
    radius,
    method='uniform'
)


# 4. Visualization
# Display Original Grayscale Image
show_image("Original Grayscale Image", gray_img, cmap='gray')

# Display LBP Texture Pattern
show_image("LBP Texture Pattern", lbp, cmap='gray')


# 5. Histogram Analysis
plt.figure(figsize=(10, 5))

plt.hist(
    lbp.ravel(),
    bins=np.arange(0, n_points + 3),
    range=(0, n_points + 2),
    color='black'
)

plt.title("LBP Histogram (Texture Distribution)")
plt.xlabel("LBP Code")
plt.ylabel("Frequency")
plt.grid(axis='y', alpha=0.3)
plt.show()
```
